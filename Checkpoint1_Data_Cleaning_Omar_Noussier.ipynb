## **Part 3: Perform a 5-Step Data Cleaning Process**


```python
# Verify the missing values
df.isnull().sum()
```

```python
# Convert 'Order_Date' to datetime format
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
# Convert 'Time' to datetime format
df['Time'] = pd.to_datetime(df['Time'])
```

```python
# Identify and Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
# Create a boxplot for dataset using Plotly
import plotly.express as px
fig = px.box(df, y=['Quantity', 'Profit'])
fig.show()
```

```python
# Identify records with unrealistic values
unrealistic_values = df[
    (df['Sales'] < 0) |
    (df['Quantity'] < 0) |
    (df['Profit'] < 0) |
    (df['Discount'] < 0) |
    (df['Shipping_Cost'] < 0)]
```

# Analysis of Extreme Values (Outliers):
From the box plot, we observe that the Quantity has very little variation, the Profit shows a significant spread with a wide interquartile range (IQR) and some extreme values extending far above the whiskers.
High-profit outliers can skew summary statistics, affecting analysis and decision-making. Proper handling involves keeping high-profit transactions for accurate insights or removing extreme values for rare anomalies.
